# Rotten Tomatoes Top 100 Movies Scraper

This project is a Python web scraper that collects the top 100 movies of all time from Rotten Tomatoes' editorial list. It extracts each movie's Tomato Meter, title and release year, and saves the cleaned data to a CSV file, ready for analysis.

Tech stack: Python, Requests, BeautifulSoup, lxml, csv

### 1. Importing Libraries

```python
import requests 
import csv 
import lxml
from bs4 import BeautifulSoup as bs
```

### 2. Defining Target URL and HTTP Headers

```python
url = "https://editorial.rottentomatoes.com/guide/best-movies-of-all-time/"
header = {
    "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/122.0.0.0 Safari/537.36 ScraperBot/1.0 (contact: test@example.com)"
}
```

### 3. Scraping, Cleaning and Exporting to CSV

```python
response = requests.get(url , headers = header)
if response.status_code == 200:
    soup = bs(response.text, "lxml")
    table = soup.find("div" , class_ = "content-body")
    for tag in table.find_all(["figure","p" , "h1" , "img"]):
        tag.decompose()
    rows = table.find_all("div")
    
    movies = []

    for row in rows:
     cols = row.find_all(["span", "a"])
     if len(cols) >= 5:
         # Use row.find() and pass the inline CSS exact string in the attrs parameter
         meter = cols[1].text.strip()
         title = cols[3].text.strip()
         year = cols[4].text.strip()

         movies.append(
                {
                    "Tomato Meter": meter,
                    "Title": title,
                    "Year": year,
                }
            )
     if len(movies) == 100 :
         break
         
    with open("top_100_movies.csv", mode="w", newline="", encoding="utf-8") as file:
        writer = csv.DictWriter(
            file, fieldnames=["Tomato Meter", "Title", "Year"]
        )
        writer.writeheader()
        writer.writerows(movies)

    print(
        f"Success! Cleaned and exported exactly {len(movies)} movies to top_100_movies.csv"
    )
else:
    print(
        f"Failed to fetch page. Status: {response.status_code}."
    )
```

### Conclusion
This project implements an end-to-end web scraping workflow in Python: sending an HTTP request with custom headers, validating the response, parsing the HTML with BeautifulSoup and lxml, removing unwanted tags, and extracting the Tomato Meter, title and year of exactly 100 movies. The cleaned data is exported to top_100_movies.csv, which can be used for further analysis with Pandas and Matplotlib.
